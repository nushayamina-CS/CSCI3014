```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
```

```python
from google.colab import files
uploaded = files.upload()

CV = list(uploaded.keys())[0]
```

# Task 1

```python
#  READ IMAGE
img = cv2.imread(CV)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Displaying IN RGB & GRAYSCALE
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.title("RGB Image")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Grayscale Image")
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.show()

#  1. DIMENSIONS
print("Image Dimensions (H,W,C):", img.shape)

#  2. PIXEL VALUES
locations = [(10,10), (50,100), (120,201)]
for (x,y) in locations:
    print(f"Pixel at {x,y} in RGB:", img[x,y])
    print(f"Pixel at {x,y} in Gray:", gray[x,y])

#  3. REGION OF INTEREST (ROI)
roi = img[50:200, 10:40]

plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.title("ROI (H 50-200, W 10-40)")
plt.axis('off')
plt.show()

#  4. DRAWING RECTANGLE
img_rect = img.copy()
cv2.rectangle(img_rect, (250, 310),(570, 600), (0, 0, 255), 5)

plt.imshow(cv2.cvtColor(img_rect, cv2.COLOR_BGR2RGB))
plt.title("Rectangle Drawn")
plt.axis('off')
plt.show()

#  5. WRITING NAME & ID
img_text = img.copy()
cv2.putText(img_text, "Nusha Yamina~231003", (300,617),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,255), 3)

plt.imshow(cv2.cvtColor(img_text, cv2.COLOR_BGR2RGB))
plt.title("Text Written")
plt.axis('off')
plt.show()
```

# Task 2

```python
#   FOR COLORED IMG
img = cv2.imread(CV)
h, w = img.shape[:2]

# CREATING BORDER
top = bottom = int(0.10 * h)
left = right = int(0.10 * w)

bordered = cv2.copyMakeBorder(
    img, top, bottom, left, right,
    cv2.BORDER_CONSTANT, value=[0, 255, 0]
)

# Displaying IMG
plt.imshow(cv2.cvtColor(bordered, cv2.COLOR_BGR2RGB))
plt.title("Image with Border")
plt.axis('off')
plt.show()

# SAVE IMG
cv2.imwrite("bordered_output.jpg", bordered)
print("Saved as bordered_output.jpg")
files.download("bordered_output.jpg")
```

```python

# FOR GREYSCALE IMG
top = bottom = int(0.10 * h)
left = right = int(0.10 * w)

# CREATING BORDER
bordered_gray = cv2.copyMakeBorder(
    gray, top, bottom, left, right,
    cv2.BORDER_CONSTANT,
    value= 0  )

# Displaying IMG
plt.imshow(bordered_gray, cmap='gray')
plt.title("Grayscale Image with Border")
plt.axis('off')
plt.show()

# SAVE IMG
cv2.imwrite("bordered_gray.jpg", bordered_gray)
print("Saved as bordered_gray.jpg")

files.download("bordered_gray.jpg")
```

```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow

def create_corner_boxes(size=500):
    # White background
    img = np.ones((size, size, 3), dtype=np.uint8) * 255

    box = size // 8

    # Top-left: Red
    cv2.rectangle(img, (0, 0), (box, box), (0, 0, 255), -1)

    # Top-right: Green
    cv2.rectangle(img, (size - box, 0), (size, box), (0, 255, 0), -1)

    # Bottom-left: Blue
    cv2.rectangle(img, (0, size - box), (box, size), (255, 0, 0), -1)

    # Bottom-right: Black
    cv2.rectangle(img, (size - box, size - box), (size, size), (0, 0, 0), -1)

    cv2_imshow(img)

create_corner_boxes(500)
```
